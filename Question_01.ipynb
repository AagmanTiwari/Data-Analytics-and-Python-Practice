### Create a small dataset, given below using commands in python

```python
# importanting dependencies
import pandas as pd
import numpy as np
```

```python
# creating data set

data = {
    'Emp_name' : ['Alice', 'Bob', 'Carol', 'Dave', 'Eve', 'Frank'],
    'Annual_Sal' : [50000, 60000, np.nan, 55000, 70000, np.nan],
    'Dept' : ['HR', 'IT', 'IT', np.nan, 'HR', 'Finance']
}

df = pd.DataFrame(data)
print("Original Data")
print(df)
```

```python
# counting missing values

print(df.isnull().sum())
```

```python
# Fill in the missing values with their respective mean values

#filling missing salary with mean
df['Annual_Sal'] = df['Annual_Sal'].fillna(df['Annual_Sal'].mean())
df['Annual_Sal']
```

```python
#filling missing text with a label

df['Dept'] = df['Dept'].fillna('Unknown')
df['Dept']
```

```python
# Display employees working in IT
df_filtered = df[df['Dept'] == 'IT']
df_filtered
```

```python
# Display employees earning more than 60000

df_filtered = df[df['Annual_Sal'] > 60000]
df_filtered
```

```python
 #calculate average salary
print(df['Annual_Sal'].mean())
```
