# A class has the following student records.Find the probability of passing given that the student is female.

### Step-01 Importing dependencies

```python
import pandas as pd
```

### Step-02 Data creation & generation

```python
data = {
    'Gender' : ['Male', 'Female','Male', 'Female','Male', 'Female','Male', 'Female' ],
    'Result' : ['Pass', 'Pass', 'Fail', 'Pass', 'Pass', 'Fail', 'Pass', 'Pass']
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# P(Pass | Female)
female = df[df['Gender'] == 'Female']
```

```python
probability = (female['Result'] == 'Pass').mean()
```

```python
print("Probability of passing given female: ", probability)
```
