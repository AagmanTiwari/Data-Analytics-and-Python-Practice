# Using the Titanic data set,calculate the probability of survival given that the passenger was female

### Step-01 Importing dependencies

```python
import seaborn as sns
import pandas as pd
```

### Step-02 Loading Dataset

```python
df = sns.load_dataset('titanic')
```

```python
df
```

### Step-03 Data Preprocessing

#### 01. Total female passengers

```python
female = df[df['sex'] == 'female']
```

#### 02. Female passengers who survived

```python
survived = female[female['survived'] == 1]
```

female

```python
survived
```

#### 03. conditional probability

```python
probability = len(survived) / len(female)
```

```python
print('P(survival|Female):', probability)
```
