# ConditionalProbability

### Question: A company collected purchase information from 100 customers. 
### Gender Purchased Not purchased 
### Male 30 20
### Female 35 15 
### Find the probability of purchase given that the customer is female.

### Step-01 importing dependencies

```python
import pandas as pd
```

### Step-02 Data creation/Generation

```python
data = {
    'Gender' : ['Male', 'Male', 'Female', 'Female'],
    'Purchased' : ['Yes', 'No', 'Yes', 'No'],
    'Count' : [30, 20, 35, 15]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

### Step-03 Data Processing and Standarization 

#### 01. Total Female Customers

```python
total_female = df[df['Gender'] == 'Female']['Count'].sum()
```

```python
print(total_female)
```

#### 02. Female customers who purchased

```python
female_purchased = df[
    (df['Gender'] == 'Female') &
    (df['Purchased'] == 'Yes')
]['Count'].sum()
```

```python
probability = female_purchased / total_female
```

```python
print("P(Purchase | Female):",probability)
```
