### Task 02 - Transformation

```python
summary = df.groupby('Department').agg(
    Max_Age = ('Age','max'),
    Low_Sal = ('Monthly_Salary', 'min'),
    Max_Emp = ('Department', 'count')
    
)
```

```python
summary
```

```python
# Filter employees by age > 30

df = df[df['Age'] > 30]
df
```

```python
# Calculate annual salary

df['Annual_Sal'] = df['Monthly_Salary'] * 12
df
```

```python
# Create a new data frame with only females, their salary, age and department

df_females = df[df['Gender'] == 'Female'][['Monthly_Salary', 'Monthly_Salary', 'Age', 'Department']]
```

```python
df_females
```

```python
# Findthe unique values of column education level
df['Education_Level'].unique()
```

```python
 # Findthe count of people belonging to each city

df.head()
df.groupby('City').agg(
    number_of_people = ('City', 'count')
)
```

```python
# Filter the male employees with salary more than 50000

df[(df['Gender'] == 'Male')&(df['Monthly_Salary'] > 50000)]
```

### Task 03 - Aggregation

```python
# Calculatetotal salary by department.

df.groupby('Department').agg(
    total_Sal = ('Monthly_Salary', 'sum')
)
```

```python
# Calculate average salary by gender

df.groupby('Gender').agg(
    avg_Sal = ('Monthly_Salary', 'mean')
)
```

```python
# Calculate average salary by city.

df.groupby('City').agg(
    avg_Sal = ('Monthly_Salary', 'mean')
)
```
